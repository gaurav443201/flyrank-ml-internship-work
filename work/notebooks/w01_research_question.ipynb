## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search question: Which pages should a content team review first for a potential refresh based on declining performance, search exposure, and content freshness?

Unit of analysis: One page/content item.

Decision: Decide which pages deserve attention first when content-review capacity is limited.

Action: A content reviewer can inspect the highest-ranked pages and decide whether to refresh, improve, consolidate, or leave the page unchanged.

Cost of a wrong recommendation: A false positive could waste reviewer time on a page that does not need a refresh. A false negative could cause a valuable page with declining performance to be overlooked. Therefore, the ranking should support human review rather than automatically prescribe an edit.
